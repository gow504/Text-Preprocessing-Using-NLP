```python
#lowercasing
df['review']=df['review'].str.lower()
```

```python
#remove html tags
def remove_html(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'',text)                 #r' ' means a raw string
```

```python
df['review']=df['review'].apply(remove_html)
```

```python
df
```

```python
#remove url's
def remove_url(text):
  pattern = re.compile(r'https?://\S+|www.\.S+')     # S+ =This is a special sequence that matches any non-whitespace character.
  return pattern.sub(r' ',text)
```

```python
df['review']=df['review'].apply(remove_url)
```

```python
df
```

```python
# remove punctuation
import string,time        # The string module contains a collection of useful string constants.
string.punctuation        # The time module provides functions for working with time,
```

```python
exclude= string.punctuation
```

```python
def remove_punc(text):
  return text.translate(str.maketrans('','',exclude)) # make trans = mapping to exclude.
                                                      # text.translate(...):This is a string method called on the text input string.
```

```python
df['review']=df['review'].apply(remove_punc)
```

```python
#remove stopwords
from nltk.corpus import stopwords
import nltk
```

```python
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
df['review'] = df['review'].apply(remove_stopwords)
```

```python
# handling Emoji's
'''
we have two options
1) replace emoji's with text
2) remove emoji's
'''

def remove_emoji(text):
    emoji_pattern = re.compile("["
                               "u\U0001F600-\U0001F64F"  # Emotions
                               "u\U0001F300-\U0001F5FF"  # Symbols & pictographs
                               "u\U0001F680-\U0001F6FF"  # Transport & map symbols
                               "u\U0001F1E0-\U0001F1FF"  # Flags(iOS)
                               "u\U00002702-\U000027B0"
                               "u\U000024C2-\U0001F251"
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
# Download the WordNet corpus
nltk.download('wordnet')          # large and structured set of texts (or speech).

# You might also need to download the 'omw-1.4' package for wordnet
nltk.download('omw-1.4')          #  OMW, collection of wordnets for various languages
```

```python
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "probably my all time favourite movie a story of selflessness sacrifice and dedication to a noble cause but its not preachy or boring it just never gets old despite my having seen some 15 or more times in the last 25 years."
punctuations = "?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
```
